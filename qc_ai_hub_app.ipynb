#### **Find an app**

```python
!qai-hub-apps list
```

Filters narrow it down, and mirror the facets on the
Apps page. Different flags are ANDed, values within one flag are ORed, and matching is case-insensitive:

- `--type` — operating system: android, windows, or ubuntu.
- `--runtime` — inference runtime: tflite, onnx, or genie.
- `--domain` and `--use-case` — for example Computer Vision and Image Classification.
- `--language` — implementation language: Python, C++, Java, Kotlin, or Go.
- `--precision` — model precision: float, w8a8, or w4a16.
- `--model` — a model the app supports, matched as a substring: mobilenet, whisper, llama.

For this walkthrough, we can narrow the catalog to the Android image classification apps that run on TFLite:

```python
!qai-hub-apps list --type android --use-case "Image Classification" --runtime tflite
```

#### **Inspect the app:**
- Before downloading anything, check what the app supports. info prints its runtime, languages, precision, the models it can bundle, the devices it has been tested on, and a link to its source on GitHub:

```python
!qai-hub-apps info image_classification_android
```

#### **Fetch the app and the Model**
- `fetch` -> downloads an app's source into a directory named after the app, in the current directory or wherever --output-dir points

- Pass a model as well and the CLI downloads it and places it exactly where the app expects to find it, so there is nothing to copy by hand. The path to the fetched app is printed when it finishes.

**Fetch the app with a model. This bundles the TFLite build of EfficientNet-B0 into the app for you:**

```python
!qai-hub-apps fetch image_classification_android --model efficientnet_b0
```

#### **Target a specific device.**
Without a target you get the app's default model build. Pass `--device` with one of the app's supported devices — or `--chipset` with a chipset — to bundle a build compiled for that hardware instead:

```python
!qai-hub-apps fetch image_classification_android --model efficientnet_b0 --device "Samsung Galaxy S25"
```

#### **Bundle a model we exported ourself.**
Point `--model-path` at exported model directory or `.zip` instead of naming a `model ID`:

```python
!qai-hub-apps fetch image_classification_android --model-path /content/efficientnet_b0
```

**Fetch the source on its own. Leave the model flags off to get just the app code, and add a model yourself later:**

```python
!qai-hub-apps fetch image_classification_android
```
